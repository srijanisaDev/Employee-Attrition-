```python
for col in ["department", "marital_status", "gender", "education", "overtime", "attrition"]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
print("Full-row duplicates:", df.duplicated().sum())
print("Duplicate employee_ids:", df["employee_id"].duplicated().sum())
```

```python
print(df["monthly_income"].value_counts().head(20))
```

```python
# are the rows with missing salary skewed toward high job levels?
df["income_missing"] = df["monthly_income"].isna()
print(df.groupby("income_missing")["job_level"].mean())
```

```python
print("promotion longer than career:", (df["years_since_last_promotion"] > df["years_at_company"]).sum())
print("tenure longer than working life:", (df["years_at_company"] > (df["age"] - 18)).sum())
```
